# Comparación de generación: prompt base y bloque de respaldo

Objetivo: reducir afirmaciones causales no respaldadas, cuantificación o magnitud inexistentes y apariencia de validación estadística, conservando un tono de marketing útil.

Tres mensajes del PR #39; LinkedIn para DEMO-001/002 y newsletter para DEMO-003. Los prompts originales se recuperan de las solicitudes registradas de las funciones de Andrea. Se mantienen mensaje, análisis y ejemplos, y se añade únicamente el bloque complementario.

Dos repeticiones por variante: 12 llamadas reales. Modelo configurado en GEMINI_MODEL (predeterminado gemini-3.5-flash-lite), temperatura 0.2 y máximo de salida 2048 tokens en ambos brazos. Las salidas anteriores son referencia histórica separada. No se prueba el endpoint ni el enrutamiento.


```python
from pathlib import Path
from comparar_prompts import cargar_casos, crear_cliente, ejecutar_comparacion

BASE = Path.cwd()
casos = cargar_casos()
adicional = (BASE / 'comparacion_api/20261020T001814Z/BLOQUE_PROBADO.txt').read_text(encoding='utf-8')
print('Casos:', [(c['id'], c['formato']) for c in casos])
```

## Solicitudes de referencia e instrucciones adicionales

Los prompts completos muestran el contexto realmente utilizado: mensaje, análisis y few-shot. Cada solicitud extendida es ese mismo texto seguido del bloque de respaldo. Las credenciales proceden de la configuración local del proceso y no forman parte del notebook.


```python
print('BLOQUE COMPLEMENTARIO:\n', adicional)
for caso in casos:
    print('\n' + '=' * 80)
    print(caso['id'], caso['formato'])
    print('ENTRADA:', caso['entrada'])
    print('PROMPT BASE COMPLETO:\n', caso['prompt_original'])
    print('RESPUESTA HISTÓRICA:\n', caso['salida_historica'])
```

## Ejecución real de la comparación

Esta celda realiza 12 solicitudes y persiste cada prompt y respuesta en Markdown. Reejecutarla consume nuevas llamadas y crea otra carpeta de ejecución. El orden se invierte en la segunda repetición. No hay reintentos ni sustitución automática de modelo.


```python
cliente = crear_cliente()
destino, resultados = ejecutar_comparacion(cliente, casos, adicional)
print('\nEvidencia guardada en:', destino)
print('Respuestas con texto:', sum(r['estado'] == 'respuesta' for r in resultados), '/', len(resultados))
```

## Contenido antes y después

Las parejas siguientes se generaron con la misma configuración. La salida base de esta comparación también es una llamada nueva, distinta de la referencia histórica. Palabras y hashtags describen formato, no veracidad o eficacia.


```python
for caso in casos:
    for rep in (1, 2):
        print('\n' + '=' * 80)
        print(caso['id'], 'REPETICIÓN', rep)
        for variante in ('base', 'extendido'):
            filas = [r for r in resultados if r['id'] == caso['id'] and r['repeticion'] == rep and r['variante'] == variante]
            if filas:
                r = filas[0]
                print('\nVARIANTE:', variante, 'ESTADO:', r['estado'])
                print(r['texto'])
                print('Palabras:', r['palabras'], 'Hashtags:', r['hashtags'])
print('\nInforme completo:', destino / 'COMPARACION_GEMINI.md')
```
